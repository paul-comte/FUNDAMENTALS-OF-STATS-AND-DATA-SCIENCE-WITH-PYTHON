3. Create a numpy array of 200 elements in which the fist 100 elements are ones; and the last 100 elements are zeros


**Solution**: we create first two arrays; one with 1 and another with the zeros, and then we will concatenate them.

```python

arr1 = np.ones(100)
arr2 = np.zeros(100)
arrFinal = np.concatenate([arr1, arr2])
print(arrFinal)
```

Alternative solution. we create an array from 0 to 199, and we impose tht all number less than 100 ar zero, and all number bigger or equal to 100 go to 1.

```python
arr = np.arange(0,200)
arr[arr<100]=0
arr[arr>=100]=1
print(arr)
```

5. open the fime `cymbals.csv` with pandas. How many measurements in the column **day1** are less than 440?

```python
df1 = pd.read_csv('data_cymbals.csv')
df1.head(2)
```

```python
data = df1['day_1']
data
```

I create a boolean array by comparing the data to the value 440. if the value in the daa array is less than 440, the boolean aray will be True, anf False otherwise.
I can then sum all the elements on the array, because True and False behaves like 1 and 0 respectively when i apply a sum

```python

bool_arr = (data<440)
print(bool_arr)
```

```python
n_meas_below_440 = sum(bool_arr)
print('there are ',n_meas_below_440,' measurement below 440 on day 1')
print('This coorrespond to', n_meas_below_440/len(data)*100,'% of the measurements')
```
